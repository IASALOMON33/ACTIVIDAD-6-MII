```python
import json

def lista_a_diccionario(persona):
    return {"nombre": persona[0], "apellido": persona[1], "edad": int(persona[2])}

def escribir_json(personas, ruta):
    with open(ruta, "w", encoding="utf-8") as f:
        json.dump(personas, f)

def leer_json(ruta):
    with open(ruta, "r", encoding="utf-8") as f:
        return json.load(f)

n = int(input("¿Cuántas personas quieres registrar? n ="))
registros = []
for i in range(n):
    print(f"Registro {i+1} de {n}")
    nombre = input("Nombre: ")
    apellido = input("Apellido: ")
    edad = input("Edad: ")
    persona_lista = [nombre, apellido, edad]
    persona_dict = lista_a_diccionario(persona_lista)
    registros.append(persona_dict)



ruta = "personas_n.json"
escribir_json(registros, ruta)
contenido = leer_json(ruta)

class Persona:
    def __init__(self, nombre, apellido, edad):
        self.nombre = nombre
        self.apellido = apellido
        self.edad = int(edad)

mi_info = Persona("Jesus", "Carrillo", 59)
print("Instancia: ", mi_info.nombre, mi_info.apellido, mi_info.edad)
```
